```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential()

model.add(Dense(16, input_shape=(3,), activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(3, activation='linear'))

model.compile(optimizer='adam', loss='mse')

states = np.array([
    [20,15,10],
    [30,5,10],
    [10,25,20],
    [15,10,30]
])

targets = np.array([
    [10,2,1],
    [10,1,2],
    [1,10,2],
    [2,1,10]
])

model.fit(states, targets, epochs=50, verbose=0)

test = np.array([[18,12,25]])

prediction = model.predict(test, verbose=0)

print("Bandwidth:", test[0])
print("Q Values:", prediction[0])
print("Selected Path:", np.argmax(prediction)+1)
```
