```python
grid = [
    ['S', '.', '.'],
    ['.', 'X', '.'],
    ['.', '.', 'G']
]

row, col = 0, 0
goal = (2, 2)
energy = 10
reward = 0

moves = ['Right', 'Right', 'Down', 'Down']

for move in moves:

    if energy == 0:
        break

    if move == 'Right' and col < 2:
        col += 1
    elif move == 'Left' and col > 0:
        col -= 1
    elif move == 'Up' and row > 0:
        row -= 1
    elif move == 'Down' and row < 2:
        row += 1

    energy -= 1

    if grid[row][col] == 'X':
        reward -= 20
        print("Obstacle Hit")
    elif (row, col) == goal:
        reward += 100
        print("Goal Reached")
        break
    else:
        reward -= 1

    print("Position:", (row, col), "Energy:", energy, "Reward:", reward)

print("\nFinal Reward:", reward)
print("Remaining Energy:", energy)
```
