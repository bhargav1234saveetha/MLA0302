```python
battery = 30
deliveries = 0
reward = 0

while battery > 0:

    if battery >= 5:
        action = "Deliver"
    else:
        action = "Stop"

    if action == "Deliver":
        battery -= 5
        deliveries += 1
        reward += 10

        print("Action:", action)
        print("Battery Left:", battery)
        print("Deliveries:", deliveries)
        print("Reward:", reward)
        print("--------------------")

    else:
        break

print("Battery Limit Reached")
print("Total Deliveries:", deliveries)
print("Final Reward:", reward)
```
