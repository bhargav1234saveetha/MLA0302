```python
grid = [
    ['S', '.', '.'],
    ['.', 'X', '.'],
    ['.', '.', 'G']
]

row = 0
col = 0

battery = 30
reward = 0

moves = ["Right", "Right", "Down", "Down"]

for move in moves:

    if battery <= 0:
        break

    if move == "Right":
        col += 1
    elif move == "Left":
        col -= 1
    elif move == "Up":
        row -= 1
    elif move == "Down":
        row += 1

    battery -= 5

    if grid[row][col] == "X":
        reward -= 20
        print("Entered No-Fly Zone")
        break

    elif grid[row][col] == "G":
        reward += 20
        print("Delivery Successful")
        break

    else:
        reward -= 1

    print("Position:", row, col)
    print("Battery:", battery)
    print("Reward:", reward)
    print("----------------")

print("Final Reward:", reward)
print("Battery Left:", battery)
```
