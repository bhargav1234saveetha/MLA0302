```python
import random

threshold = 100
consumption = 0
reward = 0

actions = ["Low", "Medium", "High"]

for step in range(10):

    action = random.choice(actions)

    if action == "Low":
        power = 10
    elif action == "Medium":
        power = 20
    else:
        power = 35

    if consumption + power <= threshold:
        consumption += power
        reward += 10
    else:
        reward -= 15

    print("Step:", step + 1)
    print("Action:", action)
    print("Consumption:", consumption)
    print("Reward:", reward)
    print("----------------------")

print("Final Consumption:", consumption)
print("Total Reward:", reward)
```
