```python
import random

roads = ["North", "South", "East", "West"]

q_values = [0, 0, 0, 0]

epsilon = 0.2

for step in range(10):

    emergency = random.choice([True, False])

    if emergency:
        action = random.randint(0, 3)
        print("Emergency Vehicle on", roads[action])
        print("Green Signal:", roads[action])

        reward = 20

    else:

        if random.random() < epsilon:
            action = random.randint(0, 3)
        else:
            action = q_values.index(max(q_values))

        print("Normal Traffic")
        print("Green Signal:", roads[action])

        reward = 5

    q_values[action] = q_values[action] + 0.1 * (reward - q_values[action])

    print("Reward:", reward)
    print("Q Values:", [round(i,2) for i in q_values])
    print("----------------------------")
```
