```python
import random

true_rewards = [0.2, 0.4, 0.8, 0.5]

trials = 500
epsilon = 0.1

print("Random Action Selection")

random_reward = 0

for i in range(trials):

    action = random.randint(0, 3)

    reward = 1 if random.random() < true_rewards[action] else 0

    random_reward += reward

print("Cumulative Reward:", random_reward)

print("\nEpsilon-Greedy Action Selection")

q = [0, 0, 0, 0]
count = [0, 0, 0, 0]

greedy_reward = 0

for i in range(trials):

    if random.random() < epsilon:
        action = random.randint(0, 3)
    else:
        action = q.index(max(q))

    reward = 1 if random.random() < true_rewards[action] else 0

    count[action] += 1

    q[action] = q[action] + (reward - q[action]) / count[action]

    greedy_reward += reward

print("Cumulative Reward:", greedy_reward)

print("\nComparison")

if greedy_reward > random_reward:
    print("Epsilon-Greedy performs better.")
else:
    print("Random Action performs better.")
```
