```python
import numpy as np
import random

true_rewards = [0.2, 0.5, 0.8, 0.4]
trials = 500

for epsilon in [0.1, 0.3, 0.5]:

    q = [0, 0, 0, 0]
    count = [0, 0, 0, 0]
    total_reward = 0

    for i in range(trials):

        if random.random() < epsilon:
            action = random.randint(0, 3)
        else:
            action = np.argmax(q)

        reward = 1 if random.random() < true_rewards[action] else 0

        count[action] += 1

        q[action] = q[action] + (reward - q[action]) / count[action]

        total_reward += reward

    print("\nEpsilon =", epsilon)
    print("Estimated Rewards:", np.round(q, 2))
    print("Machine Selected:", count)
    print("Total Reward:", total_reward)
```
